# Minimal Vector Database Demonstration

This notebook demonstrates the functionality of the `MinimalVectorDB` implemented from scratch using only NumPy. 
It covers:
1. Generating 1,000 synthetic vectors.
2. Inserting them into the database.
3. Performing a fast cosine similarity search.
4. Demonstrating CRUD operations (Delete).

```python
import numpy as np
import time
from vectordb.database import MinimalVectorDB
```

## 1. Synthetic Data Generation and Insertion
We will generate 1,000 synthetic 128-dimensional vectors and insert them into the database. 
We will also measure the time it takes to insert all vectors.

```python
# Initialize the database
db = MinimalVectorDB()

# Parameters
num_vectors = 1000
dimension = 128

# Generate synthetic vectors (simulating embeddings)
print(f"Generating {num_vectors} synthetic vectors of dimension {dimension}...")
vectors = np.random.rand(num_vectors, dimension).astype(np.float32)

# Insert vectors into the database
print("Inserting vectors into the database...")
start_time = time.time()

for i in range(num_vectors):
    db.insert(vectors[i], {"text": f"Synthetic Document {i}", "index": i})

end_time = time.time()
print(f"Successfully inserted {num_vectors} vectors in {end_time - start_time:.4f} seconds.")
```

## 2. Fast Similarity Search
We will perform a similarity search using a query vector. 
The query vector will be a slightly modified version of one of the existing vectors to ensure it finds a close match.

```python
# Create a query vector (similar to vector at index 42, with some noise)
target_index = 42
query_vector = vectors[target_index] + np.random.normal(0, 0.1, dimension).astype(np.float32)

print(f"Performing similarity search for a query similar to Document {target_index}...")
start_time = time.time()

# Search for top 5 similar vectors
results = db.search(query_vector, top_k=5)

end_time = time.time()
print(f"Search completed in {end_time - start_time:.4f} seconds.\n")

print("Top 5 Results:")
print("-" * 60)
for v_id, score in results:
    meta = db.metadata[v_id]
    print(f"ID: {v_id} | Text: {meta['text']} | Similarity Score: {score:.4f}")
print("-" * 60)
```

## 3. CRUD Operations Demonstration (Delete)
Let's verify that deleting a vector removes it from search results immediately.

```python
# Delete the top result from the previous search
top_id = results[0][0]
print(f"Deleting vector with ID: {top_id}...")
db.delete(top_id)

# Search again to verify deletion
results_after_delete = db.search(query_vector, top_k=5)
print("\nTop 5 Results after deletion:")
print("-" * 60)
for v_id, score in results_after_delete:
    meta = db.metadata[v_id]
    print(f"ID: {v_id} | Text: {meta['text']} | Similarity Score: {score:.4f}")
print("-" * 60)
```
